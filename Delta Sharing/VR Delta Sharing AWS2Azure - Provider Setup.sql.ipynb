# schema

```sql
create schema if not exists vr_demo.crisp_share
```

# sales

```sql
drop table if exists vr_demo.crisp_share.sales
```

```sql
create table vr_demo.crisp_share.sales 
tblproperties(delta.enableChangeDataFeed = true)
as select * from vr_demo.crisp.sales_raw limit 1000000
```

# store

```sql
create or replace table vr_demo.crisp_share.dim_store as select * from vr_demo.crisp.dim_store
```
