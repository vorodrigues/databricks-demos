```python
%sql
create database vr_demo.forecast
```

```python
%sql
create table vr_demo.forecast.sales_monthly as

with series as (
  select 
    product_id, 
    store_id,
    count(*) as cnt
  from vr_demo.crisp.sales_monthly
  group by all
  order by cnt desc
  limit 10
)

select sm.* from vr_demo.crisp.sales_monthly sm
inner join series s
on sm.product_id = s.product_id and sm.store_id = s.store_id
```

```python
%sql
select count(*) from vr_demo.forecast.sales_monthly
```
