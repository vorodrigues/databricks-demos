```python
import os
import json
import logging
from typing import Dict, List, Optional
# from dotenv import load_dotenv
# from aiohttp import web
# from botbuilder.core import BotFrameworkAdapterSettings, BotFrameworkAdapter, ActivityHandler, TurnContext
# from botbuilder.schema import Activity, ChannelAccount
from databricks.sdk import WorkspaceClient
from databricks.sdk.service.dashboards import GenieAPI
import asyncio
```

```python
w = WorkspaceClient()
genie_api = GenieAPI(w.api_client)
```

```python
space_id = '01efcf9d408b1c368cfd9214465aeb9d'
```

# First question

```python
question = 'qual a quantidade de itens vendidos por produto?'

if 'conversation_id' not in locals():
    conversation_id = None

if conversation_id is None:
    message = genie_api.start_conversation_and_wait(space_id, question)
    conversation_id = message.conversation_id
    message_id = message.id
else:
    message = genie_api.create_message_and_wait(space_id, conversation_id, question)
    message_id = message.id
```

```python
query_result = None
if message.query_result is not None:
    query_result = genie_api.get_message_query_result(space_id, conversation_id, message_id)
```

```python
print(query_result)
```

```python
if query_result and query_result.statement_response:
    results = w.statement_execution.get_statement(query_result.statement_response.statement_id)
```
